```python
import pandas as pd
import numpy as np
df = pd.read_csv("../Data/humanactivity.csv")
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train ,X_test ,y_train ,y_test = train_test_split(df.drop(columns = ['Activity']) , df['Activity'] , test_size=0.2) 
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis


lda = LinearDiscriminantAnalysis(n_components=2)
```

```python
X_train_lda = lda.fit_transform(X_train , y_train)
```

```python
lda_df = pd.DataFrame(X_train_lda)
lda_df['activity'] = y_train
```

```python
lda_df.head()
```

```python
import seaborn as sns


sns.scatterplot(data = lda_df , x= lda_df[0] , y = lda_df[1] , hue = 'activity')
```
