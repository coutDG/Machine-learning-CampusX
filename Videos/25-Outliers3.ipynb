```python
import pandas as pd
import numpy as np
df = pd.read_csv("../Data/Placement.csv")
```

```python
df.head()
```

```python
import seaborn as sns
sns.boxplot(df['placement_exam_marks'])
```

```python
upper = df['placement_exam_marks'].quantile(0.99)
lower = df['placement_exam_marks'].quantile(0.01)
```

```python
new_df = df[~((df['placement_exam_marks'] > upper) | (df['placement_exam_marks'] <lower)) ]
```

```python
new_df['placement_exam_marks'].describe()
```

```python
df['placement_exam_marks'].describe()
```

```python
sns.boxplot(new_df['placement_exam_marks'])
```

```python
df['placement_exam_marks'] = np.where(df['placement_exam_marks'] > upper  , upper , np.where(df['placement_exam_marks'] <lower , lower , df['placement_exam_marks'] ))
```

```python
df.shape
```

```python
sns.boxplot(df['placement_exam_marks'])
```

```python
df['placement_zscore'] = (df['placement_exam_marks']  - df['placement_exam_marks'].mean()) / df['placement_exam_marks'].std()
```

```python
df[df['placement_zscore'].abs() > 3]
```
