```python
import plotly.graph_objects as go
import plotly.express as px 
import pandas as pd
df = px.data.gapminder()
```

```python
df.head()
```

```python
df_grouped = df.groupby(['year' , 'continent']).sum(numeric_only = True).reset_index()
```

```python
asia = df[df['continent'] == 'Asia'].groupby('year', as_index=False)['pop'].sum()
europe = df[df['continent'] == 'Europe'].groupby('year', as_index=False)['pop'].sum()

fig = go.Figure(data = [go.Bar(name='Asia', x=asia['year'], y=asia['pop']) , go.Bar(name='Europe', x=europe['year'], y=europe['pop'])])
fig.update_layout(barmode = 'stack')
fig.show()
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x = df[df['country'] == "United States"].year , y = df[df['country']== 'United States'].lifeExp )) 
fig.add_trace(go.Scatter(x = df[df['country'] == "Japan"].year , y = df[df['country']== 'Japan'].lifeExp ))
fig.show() 
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x = df[df['country'] == "United States"].year , y = df[df['country']== 'United States'].lifeExp )) 
fig.add_trace(go.Scatter(x = df[df['country'] == "Japan"].year , y = df[df['country']== 'Japan'].lifeExp  , name = 'japan' , line = dict(color = 'sienna' , width = 5)))
fig.show()  
```
