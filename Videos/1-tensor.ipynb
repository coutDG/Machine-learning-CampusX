```python
import numpy as np 
```

## 0D Tensors 
- Basically Scalars 

```python
a = np.array(5)
print(a.ndim)
print(a.shape)
print(a)
```

## 1D Tensors
- Vectors
- Still can have multiple dimensions

```python
a = np.array([1,2,3])
print(a.ndim)
print(a.shape)
print(a)
```

## 2D Tensors
- Collections of vectors(matrix)
- 2 Dimensional Tensors ( Row and column )

```python
a = np.array([[1,2,3],[3,4,5],[5,3,2]])
print(a.ndim)
print(a.shape)
print(a)
```

## 3D Tensors
- Collection of matrices
- 3D(row column depth)

```python
a = np.array([[[1,2,3]],[[2,3,4]]])
print(a.ndim)
print(a.shape)
print(a)
```
