```python
import plotly.express as px 
df = px.data.gapminder()
```

```python
df
```

```python
x = df.groupby("year")['gdpPercap'].sum()
```

```python
fig = px.line(x)
fig.show()
```

```python
new_df = df[(df['continent'] == 'Europe') & (df['year'] == 2007)]
```

```python
fig = px.scatter(new_df , x = 'gdpPercap' , y = 'lifeExp' , size = 'pop' , color = 'country' , hover_data = 'iso_alpha' , hover_name = 'country')
fig.show()
```

```python
df_1990 = df[(df['continent'] == 'Europe') & (df['year'] > 1990)]
```

```python
fig = px.scatter(df_1990 , x = 'gdpPercap' , y = 'lifeExp' , facet_row = "year" , color = 'country' , height = 500)
```

```python
fig.show()
```

```python
df_new = df[df['year'] > 1990]
```

```python
df_new['continent'].value_counts()
```

```python
fig  = px.scatter(df_new , x = 'gdpPercap' , y ='lifeExp' , color = 'country' , facet_row = 'year' , facet_col = 'continent' , height  = 500)
fig.show()
```

```python
oceania_df = df[df['continent'] == 'Oceania']
fig = px.line(oceania_df , x = 'year' , y = 'lifeExp' , color = 'country' , text = 'lifeExp' , template = 'seaborn')
fig.update_traces(textposition = 'bottom right')
fig.update_layout(title = "Oceania Race")
fig.show()
```

```python
year_con_df = df.groupby(['year' , 'continent']).sum(numeric_only = True).reset_index()
```

```python
fig = px.bar(year_con_df , x = "continent" , y = 'pop' , color = 'continent' , template = 'ggplot2' , facet_col='year' , facet_col_wrap=4)
fig.show()
```
