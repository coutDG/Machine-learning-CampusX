```python
np.random.seed(1)
n = 1000

# lognormal gives you the right skew + high-earner tail naturally
salary = np.random.lognormal(mean=10.5, sigma=0.6, size=n)

df = pd.DataFrame({
    'employee_id': range(n),
    'salary': salary,
    'years_experience': np.random.randint(0, 30, n)
})

missing_idx = np.random.choice(df.index, size=int(0.08 * n), replace=False)
df.loc[missing_idx, 'salary'] = np.nan

```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
trf = ColumnTransformer([
    ("trf1" , SimpleImputer(strategy="mean" ) , ['salary'])
] ,remainder="passthrough")
```

```python
X_transformed = trf.fit_transform(df)

new_df = pd.DataFrame(X_transformed , columns = ['salary','employee_id'  ,'yoe'])
```

```python
np.random.seed(7)
n = 1000

age = np.random.randint(18, 80, n)
height = np.random.normal(165, 10, n) + (age < 25) * -5  

weight = 0.9 * height - 0.15 * age + np.random.normal(0, 5, n) - 60

df = pd.DataFrame({'age': age, 'height': height, 'weight': weight})

missing_idx = np.random.choice(df.index, size=int(0.15 * n), replace=False)
df.loc[missing_idx, 'weight'] = np.nan

print(df.corr())
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import KNNImputer

knn_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('imputer', KNNImputer(weights='distance', n_neighbors=5))
])

trf = ColumnTransformer([
    ("knn", knn_pipeline, ['age', 'height', 'weight'])
], remainder="passthrough")
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
iter_pipeline = Pipeline([
    ("scaler" , StandardScaler()),
    ("imputer" , IterativeImputer(max_iter=10 , random_state=0))
])


trf = ColumnTransformer([
    ("iter" , iter_pipeline , ['age' ,'height' ,'weight'])
] ,remainder="passthrough")
```
