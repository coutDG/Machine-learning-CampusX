```python
import pandas as pd
import seaborn as sns
```

```python
df = sns.load_dataset("titanic")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.drop(columns = ['class' , 'who', 'adult_male' ,'deck' , 'embark_town' , 'alive' , 'alone'] , inplace = True)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
```

```python
X_train , X_test , y_train ,y_test = train_test_split(df.drop(columns = ['survived']) , df['survived'] ,test_size=0.2 , random_state=42)
```

```python
from sklearn.preprocessing import OneHotEncoder


embarked_pipeline = Pipeline([
    ("impute" , SimpleImputer(strategy="most_frequent")),
    ("one_hot" , OneHotEncoder(handle_unknown="ignore"))
])

col_transformer = ColumnTransformer(
    transformers=[
        ("imputer_age" , SimpleImputer() , ['age']),
        ("impute_oneHot_embarked" , embarked_pipeline , ['embarked']),
        ("one_hot_sex" , OneHotEncoder(handle_unknown="ignore") , ['sex'])
        ], remainder="passthrough"
)
```

```python
from sklearn.preprocessing import StandardScaler


full_pipeline = Pipeline([
    ("preprocessor" , col_transformer),
    ("scaler" , StandardScaler())
])
```

```python
X_train_transformed = full_pipeline.fit_transform(X_train)
```

```python
X_test_transformed = full_pipeline.transform(X_test)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
clf = DecisionTreeClassifier()
clf.fit(X_train_transformed,  y_train)
y_pred = clf.predict(X_test_transformed)

print(accuracy_score(y_test ,y_pred))
```
