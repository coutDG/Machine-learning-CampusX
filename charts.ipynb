```python
plt.bar(df['sex'] , df['tip']) # shows the average tips from both genders
```

```python
plt.barh(df['sex'] , df['tip'])
```

```python
plt.scatter(df['total_bill'] , df['tip'])
```

```python
plt.hist(df['total_bill'] , bins = 20 )
plt.show()
```

```python
plt.boxplot(df['tip'])
plt.show()
```

```python
x  = df.groupby("day")['total_bill'].sum()
plt.pie(x, autopct = "%.2f%%" , labels = ["thur", 'fri', 'sat','sun' ])
plt.show()
```

```python
fig , axes = plt.subplots(ncols = 2,figsize = (12,5))

axes[0].scatter(df['total_bill'],  df['tip'])
axes[0].set_title("total_bill vs tip")
axes[0].set_xlabel("total_bill")
axes[0].set_ylabel("tip")


axes[1].hist(df['total_bill'])
axes[1].set_title("Totalbill histogram")
axes[1].set_xlabel("total_bill")
```
