```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("../Data/salary_data.csv")
```

```python
df.head()
```

```python
class LinearRegression:
    def __init__(self):
        self.m = None
        self.b = None
        
    def fit(self,  X_train ,y_train):
        X_train = np.asarray(X_train).ravel()
        y_train = np.asarray(y_train).ravel()
        
        numerator = 0
        denominator = 0
        
        x_mean = X_train.mean()  
        y_mean = y_train.mean()
        
        for i in range(X_train.shape[0]):
            numerator += (X_train[i] - x_mean)*(y_train[i] - y_mean)
            denominator += (X_train[i] - x_mean)**2
            
        self.m = numerator/denominator
        self.b = y_mean - self.m * x_mean
    
    def predict(self , X_test):
        X_test = np.asarray(X_test).ravel()
        return self.m*X_test + self.b
```

```python
X_train , X_test , y_train ,y_test = train_test_split(df['YearsExperience'], df['Salary'] , test_size=0.2)
```

```python
lr = LinearRegression()
lr.fit(X_train , y_train)
y_pred = lr.predict(X_test)

print(y_pred)
```
